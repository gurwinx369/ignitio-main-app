## Step 1: Create a Workspace

First, we need a directory to work in. Think of this as your project folder.

````python
import os
from pathlib import Path

# Create a practice directory
workspace = Path('/tmp/git-practice')
workspace.mkdir(parents=True, exist_ok=True)
os.chdir(workspace)

print(f"✅ Workspace created!")
print(f"📁 Location: {workspace}")
print(f"📍 Current directory: {os.getcwd()}")
````

---

## Step 2: Initialize Git Repository

The `git init` command transforms a regular folder into a Git repository.

### What happens when you run `git init`?
- Creates a hidden `.git` folder
- Sets up the repository structure
- Prepares for version control

**Try it now:**

````python
# Initialize a new Git repository
!git init
````

### 🔍 Let's verify the `.git` folder was created:

````python
# List all files including hidden ones
!ls -la
````

**Expected output:** You should see a `.git` directory in the list!

---

## Step 3: Configure Git Identity

Every Git commit needs to know who made it. Let's set your identity.

### Why is this important?
- Tracks who made each change
- Required for commits
- Shows up in `git log`

````python
# Configure your name (replace with your actual name)
!git config user.name "Alex Developer"

# Configure your email (replace with your actual email)
!git config user.email "alex@example.com"

print("\n✅ Git identity configured!")
print("\n📋 Your Git configuration:")
!git config --list | grep user
````

---

## Step 4: Understanding Git Status

The `git status` command is your best friend. It shows:
- Which branch you're on
- Which files have changed
- Which files are staged for commit
- Which files are untracked

````python
# Check repository status
!git status
````

**Current status:**
- ✅ On branch `main` (or `master`)
- ✅ No commits yet
- ✅ Nothing to commit

This is expected for a fresh repository!
