# C1 — Remote Rider

**One-line goal:** Create GitHub repo, add remote, push initial commits, configure SSH/HTTPS.

```python
%%bash
set -e
git remote add origin git@github.com:youruser/yourrepo.git
git push -u origin main
ssh-keygen -t ed25519 -C "you@example.com"  # follow prompts
ssh -T git@github.com  # test auth
```

### Essential notes

- Use SSH for convenience once keys are set up; HTTPS is fine with PATs.
- Set `main` (or your preferred default) and push with `-u` to set upstream.
