# A2 — First Repo Forge

**One-line goal:** Create a local repo (git init), inspect .git, perform add → commit flow.

```python
%%bash
set -e
mkdir my-sandbox && cd my-sandbox
git init
ls -la .git
echo "# My Project" > README.md
git add README.md
git commit -m "chore: add README"
```

### Essential notes

- `.git` is the hidden directory storing the repo database and objects.
- Keep commits small and meaningful for easier history navigation.
